```python
def get_env_configs():
    ''''
    cdp 环境使用 PatchGroup 作为区分环境的tag, 且只有 UAT 和 PROD 环境
    UAT环境: PatchGroup值为 NONPROD
    PROD环境: PatchGroup值为 PROD
    '''
    json_str = spark.conf.get("spark.databricks.clusterUsageTags.clusterAllTags")
    data = json.loads(json_str)

    env = ""
    # 查找 key 为 PatchGroup 的值
    for item in data:
        if item['key'].upper() == 'PATCHGROUP':
            env = item['value'].upper()
            break

    if env == "NONPROD":
        return uat_config
    elif env == "PROD":
        return prd_config
    else:
        return None



def get_kv(value, kv_scope= None):
    '''
    value: kv格式配置值. 例 AK:MysqlPassword
    '''
    match = re.match(r'(?i)AK:(.*)', value)
    if match:
        kv_key= match.group(1)        
        #print(kv_scope+"   "+kv_key)

        if kv_scope == None:
          error_msg=f"{kv_scope} is null,please check {value}!!!"
          raise Exception(error_msg)
        else:
          return  dbutils.secrets.get(kv_scope,kv_key)
    else:
        return value


#判断空
def check_empty(obj):
  if isinstance(obj, str):
      obj = obj.strip()
  if not obj:
    return True
  else:
    return False


def get_env_config(key_path,null_chek=False):

    env_configs = get_env_configs()
    config_dict = json.loads(json.dumps(env_configs))
    kv_scope = config_dict.get("kv_scope")

    # 通过键路径获取对应的值
    keys = key_path.split('.')
    value = config_dict
    for key in keys:
        value = value.get(key)
        if value is None:
            break
    
    if(null_chek and check_empty(value)):
      error_msg=f"{key_path} is null,please check it!!!"
      raise Exception(error_msg)

    #print(f"Get value path:{key_path},value:{value}")

    if isinstance(value, str):
        value = get_kv(value, kv_scope)

    return value
```
