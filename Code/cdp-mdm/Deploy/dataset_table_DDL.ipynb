```python
%sql
create schema if not exists catalog_southeastasia_mdm_golden_prod.consumer_combine;
```

```python
%sql
use catalog_southeastasia_mdm_golden_prod.consumer_combine
```

```python
%sql
CREATE OR REPLACE TABLE t_cbr_dataset (
  `MarketCode` STRING,
  `MDMKey` STRING,
  `FinalJSON` STRING,
  `CreatedTimestamp` TIMESTAMP,
  `UpdatedTimestamp` TIMESTAMP,
  `TASK_ID` STRING
)
USING delta
TBLPROPERTIES (
  'delta.enableChangeDataFeed' = 'true'
)
```

```python
%sql

CREATE OR REPLACE TABLE  t_transaction_master_dataset (
    tran_id STRING,
    tran_mrkt_code STRING,
    tran_brnd_code STRING,
    tran_srcs_code STRING,
    tran_order_id STRING,
    json_str STRING,
    task_id STRING,
    creation_dt TIMESTAMP,
    update_dt TIMESTAMP
)
USING delta
TBLPROPERTIES (
  'delta.enableChangeDataFeed' = 'true'
)
```
