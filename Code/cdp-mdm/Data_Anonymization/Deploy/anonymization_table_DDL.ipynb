# Data Anonymization Tables

```python
%sql
create schema if not exists catalog_southeastasia_mdm_silver_uat.data_anonymization
```

```python
%sql
CREATE OR REPLACE TABLE catalog_southeastasia_mdm_silver_uat.data_anonymization.t_mdm_anonymization_log (
  cal_uuid STRING,
  ConsumerId STRING,
  MarketCode STRING,
  BrandCode STRING,
  SourceSystemCode STRING,
  Old_UniversalKey STRING,
  New_UniversalKey STRING,
  LastActivityTime TIMESTAMP,
  status STRING COMMENT 'InProgress | Complete',
  comment STRING,
  create_time TIMESTAMP,
  task_id STRING
)
USING DELTA
COMMENT 'Consumer Deletion / Anonymization 处理日志'
```

```python
%sql
CREATE OR REPLACE TABLE catalog_southeastasia_mdm_silver_uat.data_anonymization.t_cbr_dataset_backup (
  MarketCode STRING,
  MDMKey STRING,
  FinalJSON STRING,
  cbr_type STRING COMMENT 'cbr | cbrpublic | cbrdj',
  create_time TIMESTAMP,
  expired_time TIMESTAMP,
  task_id STRING
)
USING DELTA
COMMENT 'CBR dataset backup for anonymized consumers'
```
